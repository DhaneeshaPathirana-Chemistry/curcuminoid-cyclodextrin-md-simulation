```python
# Cell 2: Setup working directory
import os
import shutil

structures_path = "/content/drive/MyDrive/curcuminoid_cyclodextrin_project/structures"
docking_path = "/content/drive/MyDrive/curcuminoid_cyclodextrin_project/Docking Results"
work_dir = "/content/docking"

os.makedirs(work_dir, exist_ok=True)
os.makedirs(docking_path, exist_ok=True)
os.chdir(work_dir)

print(f"✅ Working directory: {os.getcwd()}")
```

```python
# Cell 3: Check if files are in Drive
print("📁 Checking files in structures/...")
print(os.listdir(structures_path))
```

```python
# Cell 4: Copy files from Drive
print("📁 Copying files...")

shutil.copy(f"{structures_path}/sbe_beta_cd_receptor.pdbqt", "receptor.pdbqt")
shutil.copy(f"{structures_path}/bdmc_minimized by PyRx_for docking.pdbqt", "ligand.pdbqt")

print("   ✅ Files copied")
print("📁 Files in working directory:")
print(os.listdir())
```

```python
# Cell 5: Install AutoDock Vina
print("📦 Installing AutoDock Vina...")
!apt-get update -qq
!apt-get install -y autodock-vina -qq
print("   ✅ Vina installed!")
```

```python
# Cell 6: Run docking
print("🎬 Running docking...")

!vina --receptor receptor.pdbqt \
     --ligand ligand.pdbqt \
     --center_x 3.34 --center_y -0.60 --center_z -2.69 \
     --size_x 18 --size_y 18 --size_z 18 \
     --exhaustiveness 16 --num_modes 10 \
     --out bdmc_sbe_docked18.pdbqt > bdmc_sbe_log18.txt 2>&1

print("   ✅ Docking complete!")
```

```python
# Cell 7: Show results
print("\n" + "="*60)
print("📊 DOCKING RESULTS: bdmc18 + sbe_β-CD")
print("="*60)

with open("bdmc_sbe_log18.txt", "r") as f:
    print(f.read())
```

```python
# Cell 8: Save results to Drive
print("💾 Saving results...")

if os.path.exists("bdmc_sbe_docked18.pdbqt"):
    shutil.copy("bdmc_sbe_docked18.pdbqt", f"{docking_path}/bdmc_sbe_docked18_colab.pdbqt")
    shutil.copy("bdmc_sbe_log18.txt", f"{docking_path}/bdmc_sbe_log18_colab.txt")
    print(f"✅ Results saved to: {docking_path}")
else:
    print("❌ bdmc_sbe_docked18.pdbqt not found")
```

```python
# Cell 6: Run docking
print("🎬 Running docking...")

!vina --receptor receptor.pdbqt \
     --ligand ligand.pdbqt \
     --center_x 3.34 --center_y -0.60 --center_z -2.69 \
     --size_x 14 --size_y 14 --size_z 14 \
     --exhaustiveness 16 --num_modes 10 \
     --out bdmc_sbe_docked14.pdbqt > bdmc_sbe_log14.txt 2>&1

print("   ✅ Docking complete!")
```
