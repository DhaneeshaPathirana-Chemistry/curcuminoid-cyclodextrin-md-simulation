```python
# Cell 2: Setup working directory
import os
import shutil

structures_path = "/content/drive/MyDrive/curcuminoid_cyclodextrin_project/structures"
docking_path = "/content/drive/MyDrive/curcuminoid_cyclodextrin_project/Docking Results"
work_dir = "/content/docking"

os.makedirs(work_dir, exist_ok=True)
os.makedirs(docking_path, exist_ok=True)
os.chdir(work_dir)

print(f"✅ Working directory: {os.getcwd()}")
```

```python
# Cell 3: Check files in Drive
print("📁 Checking files in structures/...")
print(os.listdir(structures_path))
```

```python
# Cell 4: Copy files to working directory
print("📁 Copying files...")

shutil.copy(f"{structures_path}/bcd_minimized_byAutodock.pdbqt", "receptor.pdbqt")
shutil.copy(f"{structures_path}/curcumin_minimized_byAutodock.pdbqt", "ligand.pdbqt")

print("   ✅ Files copied")
print("📁 Files in working directory:")
print(os.listdir())
```

```python
# Cell 5: Install AutoDock Vina
print("📦 Installing AutoDock Vina...")
!apt-get update -qq
!apt-get install -y autodock-vina -qq
print("   ✅ Vina installed!")
```

```python
# Cell 6: Run docking with 14 Å grid box
print("🎬 Running docking with 14 Å grid box...")

!vina --receptor receptor.pdbqt \
     --ligand ligand.pdbqt \
     --center_x -0.01 --center_y -0.00 --center_z 0.01 \
     --size_x 14 --size_y 14 --size_z 14 \
     --exhaustiveness 16 --num_modes 10 \
     --out cur_bcd_docked.pdbqt > cur_bcd_log.txt 2>&1

print("   ✅ Docking complete!")
```

```python
# Cell 7: Show results
print("\n" + "="*60)
print("📊 DOCKING RESULTS (14 Å GRID)")
print("="*60)

with open("cur_bcd_log.txt", "r") as f:
    print(f.read())
```

```python
# Cell 8: Save results to Drive
print("💾 Saving results...")

if os.path.exists("cur_bcd_docked.pdbqt"):
    shutil.copy("cur_bcd_docked.pdbqt", f"{docking_path}/cur_bcd_docked_14A.pdbqt")
    shutil.copy("cur_bcd_log.txt", f"{docking_path}/cur_bcd_log_14A.txt")
    print(f"✅ Results saved to: {docking_path}")
else:
    print("❌ Docking failed. Check Cell 7 output.")
```

```python
print("🎬 Running docking with 16 Å grid box...")

!vina --receptor receptor.pdbqt \
     --ligand ligand.pdbqt \
     --center_x -0.01 --center_y -0.00 --center_z 0.01 \
     --size_x 16 --size_y 16 --size_z 16 \
     --exhaustiveness 16 --num_modes 10 \
     --out cur_bcd_docked_16A.pdbqt > cur_bcd_log_16A.txt 2>&1

print("   ✅ Docking complete!")
```
